```python
# CLAUDE hms-best-attempt: offline late submission (internet OFF).
# Inputs: competition data + notebook outputs that contain repo/ and runs/<EXP>/{config.yaml, fold*.pt}.
# A Kaggle notebook-output input is pinned to ONE version, so models from different versions of the training notebook
# come in through different notebooks (e.g. training V14 -> FA/FB, weights-store notebook -> D).
import glob, os, subprocess
W = {'FB': 0.5, 'FA': 0.5}          # experiment -> blend weight (set from the cross-fitted OOF blend)
R = os.path.dirname(os.path.dirname(glob.glob('/kaggle/input/**/repo/hmsbest/infer.py', recursive=True)[0]))
D = os.path.dirname([p for p in glob.glob('/kaggle/input/**/test.csv', recursive=True) if 'hms' in p][0])
exps = []
for name, w in W.items():
    c = sorted(glob.glob(f'/kaggle/input/**/runs/{name}/config.yaml', recursive=True))
    assert c, f'no runs/{name} in the inputs'
    exps.append(f'--exp {os.path.dirname(c[0])}:{w}')
print(R, D, *exps, sep='\n')
subprocess.run(f'cd {R} && python -m hmsbest.infer --data_root {D} {" ".join(exps)} --out /kaggle/working/submission.csv',
               shell=True, check=True)
```

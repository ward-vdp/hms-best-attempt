```python
# CLAUDE hms-best-attempt: weights store (CPU, no GPU needed). Input: the training notebook output of the version that holds
# the wanted runs. Copies those runs (and the code) into this notebook's output so a submission can combine them with
# runs from another version of the training notebook.
import glob, os, shutil
KEEP = ['D']
src = os.path.dirname(os.path.dirname(glob.glob('/kaggle/input/**/runs/D/config.yaml', recursive=True)[0]))
for name in KEEP:
    shutil.copytree(f'{src}/runs/{name}', f'/kaggle/working/runs/{name}')
if os.path.exists(f'{src}/repo'):
    shutil.copytree(f'{src}/repo', '/kaggle/working/repo', dirs_exist_ok=True)
print(os.popen('du -sh /kaggle/working/runs/* && ls /kaggle/working/runs/D').read())
```

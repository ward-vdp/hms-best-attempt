```python
# CLAUDE hms-best-attempt: weights store + CV-definition diagnostic (CPU; internet on for git clone).
# Input: the training-runner output of the version that holds the wanted runs (pinned). Copies those runs and the latest code
# into this output, so a submission can combine them with runs from another version of the runner.
import glob, os, shutil, subprocess
src = os.path.dirname(os.path.dirname(sorted(glob.glob('/kaggle/input/**/runs/pair_*', recursive=True))[0]))
print(src)
_ = [shutil.copytree(f'{src}/runs/{n}', f'/kaggle/working/runs/{n}') for n in ['D'] if os.path.exists(f'{src}/runs/{n}')]
subprocess.run('git clone -q https://github.com/ward-vdp/hms-best-attempt /kaggle/working/repo', shell=True, check=True)
T = [p for p in glob.glob('/kaggle/input/**/train.csv', recursive=True) if 'hms' in p][0]
subprocess.run(f"cd /kaggle/working/repo && python -m hmsbest.cvdef --train {T} --rows '{src}/runs/pair_*/rows_for*.csv'", shell=True, check=True)
print(os.popen('du -sh /kaggle/working/*').read())
```
